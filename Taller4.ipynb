```python
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt
import seaborn as sns

# Datos proporcionados
media_muestral = 495  # Media de la muestra
mu_0 = 500  # Media poblacional hipotetizada (según H0)
desviacion_estandar_poblacional = 15  # Desviación estándar poblacional
n = 40  # Tamaño de la muestra
alfa = 0.05  # Nivel de significancia

# 1. Defina las hipótesis nula y alternativa.
# H0: mu = 500
# Ha: mu < 500 (Prueba de cola izquierda)

# 2. Establezca el nivel de significancia α.
# alfa = 0.05

# 3. Identifique el estadístico de prueba (Z) y su distribución.
# Z = (x_bar - mu_0) / (sigma / sqrt(n))
# Distribución: Normal estándar

# 4. Determine el valor crítico y la región de rechazo.
# Para una prueba de cola izquierda con alfa = 0.05, el valor crítico Z es el valor tal que P(Z < valor_critico_z) = 0.05.
valor_critico_z = norm.ppf(alfa)

# Región de rechazo: Z < valor_critico_z

# 5. Calcule el estadístico de prueba y el valor p.
error_estandar = desviacion_estandar_poblacional / np.sqrt(n)
estadistico_z = (media_muestral - mu_0) / error_estandar

# El p-valor para una prueba de cola izquierda es la probabilidad de obtener un valor del estadístico Z igual o menor al calculado.
p_valor = norm.cdf(estadistico_z)

print(f"Estadístico de prueba Z: {estadistico_z:.4f}")
print(f"Valor crítico de Z (alfa={alfa}): {valor_critico_z:.4f}")
print(f"Valor p: {p_valor:.4f}")

# 6. Grafique la distribución correspondiente, indicando claramente la región de rechazo, el valor crítico y el estadístico calculado.

# Rango de valores para graficar la distribución normal estándar
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x, 0, 1) # PDF de la distribución normal estándar

plt.figure(figsize=(10, 6))
plt.plot(x, y, label='Distribución Normal Estándar')

# Rellenar la región de rechazo (cola izquierda)
x_rechazo = np.linspace(-4, valor_critico_z, 100)
y_rechazo = norm.pdf(x_rechazo, 0, 1)
plt.fill_between(x_rechazo, y_rechazo, color='red', alpha=0.5, label=f'Región de Rechazo (α = {alfa})')

# Marcar el valor crítico de Z
plt.axvline(valor_critico_z, color='red', linestyle='--', label=f'Valor Crítico Z = {valor_critico_z:.4f}')

# Marcar el estadístico de prueba Z calculado
plt.axvline(estadistico_z, color='blue', linestyle='-', label=f'Estadístico Z = {estadistico_z:.4f}')

# Configuración del gráfico
plt.title('Distribución Normal Estándar con Región de Rechazo y Estadístico de Prueba')
plt.xlabel('Valor Z')
plt.ylabel('Densidad de Probabilidad')
plt.legend()
plt.grid(True)
plt.show()

# 7. Tome una decisión estadística y concluya en términos prácticos, explicando el significado del valor p.

# Decisión estadística
print("\nDecisión estadística:")
if estadistico_z < valor_critico_z:
    print(f"El estadístico de prueba Z ({estadistico_z:.4f}) es menor que el valor crítico Z ({valor_critico_z:.4f}).")
    print("Por lo tanto, rechazamos la hipótesis nula (H0).")
elif p_valor < alfa:
     print(f"El p-valor ({p_valor:.4f}) es menor que el nivel de significancia ({alfa}).")
     print("Por lo tanto, rechazamos la hipótesis nula (H0).")
else:
    print(f"El estadístico de prueba Z ({estadistico_z:.4f}) no es menor que el valor crítico Z ({valor_critico_z:.4f}).")
    print(f"El p-valor ({p_valor:.4f}) no es menor que el nivel de significancia ({alfa}).")
    print("Por lo tanto, no rechazamos la hipótesis nula (H0).")

# Conclusión en términos prácticos y significado del valor p
print("\nConclusión en términos prácticos:")
if estadistico_z < valor_critico_z:
    print(f"Con un nivel de significancia del {alfa*100}%, hay suficiente evidencia para concluir que el peso promedio de los paquetes de cereal es significativamente menor a 500 g.")
    print(f"\nSignificado del valor p ({p_valor:.4f}):")
    print(f"El p-valor es la probabilidad de obtener una media muestral de 495 g o menos (si la hipótesis nula es verdadera y la media poblacional es 500 g). Un p-valor pequeño ({p_valor:.4f}) indica que es muy poco probable observar un resultado como el de nuestra muestra si la hipótesis nula fuera cierta. Dado que el p-valor ({p_valor:.4f}) es menor que nuestro nivel de significancia ({alfa}), rechazamos H0.")
else:
    print(f"Con un nivel de significancia del {alfa*100}%, no hay suficiente evidencia para concluir que el peso promedio de los paquetes de cereal es significativamente menor a 500 g.")
    print("No podemos rechazar la afirmación de que el peso promedio es de 500 g.")
    print(f"\nSignificado del valor p ({p_valor:.4f}):")
    print(f"El p-valor es la probabilidad de obtener una media muestral de 495 g o menos (si la hipótesis nula es verdadera y la media poblacional es 500 g). Un p-valor grande ({p_valor:.4f}) indica que es razonablemente probable observar un resultado como el de nuestra muestra si la hipótesis nula fuera cierta. Dado que el p-valor ({p_valor:.4f}) no es menor que nuestro nivel de significancia ({alfa}), no rechazamos H0.")
```

```python
import numpy as np
from scipy.stats import chi2 # Importamos la distribución chi-cuadrado
import matplotlib.pyplot as plt
import seaborn as sns

# Datos proporcionados
varianza_muestral = 0.3  # Varianza de la muestra (s^2)
sigma2_0 = 0.5  # Varianza poblacional hipotetizada (según H0, sigma^2)
n = 12  # Tamaño de la muestra
alfa = 0.05  # Nivel de significancia

# 1. Defina las hipótesis nula y alternativa.
# H0: sigma^2 = 0.5 (La varianza del grosor es de 0.5 mm^2)
# Ha: sigma^2 < 0.5 (La varianza del grosor es menor a 0.5 mm^2 - Prueba de cola izquierda)

# 2. Establezca el nivel de significancia α.
# alfa = 0.05

# 3. Identifique el estadístico de prueba (chi^2) y su distribución.
# chi^2 = ((n - 1) * s^2) / sigma^2
# Distribución: Chi-cuadrado con n-1 grados de libertad

# 4. Determine los grados de libertad y el valor crítico, además de la región crítica y la región de rechazo.
grados_libertad = n - 1

# Para una prueba de cola izquierda con alfa = 0.05 y n-1 grados de libertad, el valor crítico chi^2 es el valor tal que P(Chi^2 < valor_critico_chi2) = 0.05.
valor_critico_chi2 = chi2.ppf(alfa, df=grados_libertad)


# Región de rechazo: chi^2 < valor_critico_chi2

# 5. Calcule el estadístico de prueba y el valor p.
estadistico_chi2 = ((n - 1) * varianza_muestral) / sigma2_0

# El p-valor para una prueba de cola izquierda es la probabilidad de obtener un valor del estadístico chi^2 igual o menor al calculado.
p_valor = chi2.cdf(estadistico_chi2, df=grados_libertad) # Usamos cdf para la cola izquierda


print(f"Estadístico de prueba Chi^2: {estadistico_chi2:.4f}")
print(f"Grados de libertad: {grados_libertad}")
print(f"Valor crítico de Chi^2 (alfa={alfa}): {valor_critico_chi2:.4f}")
print(f"Valor p: {p_valor:.4f}")

# 6. Grafique la distribución correspondiente, indicando claramente la región de rechazo, el valor crítico y el estadístico calculado.

# Rango de valores para graficar la distribución Chi-cuadrado
# Ajustamos el rango para que sea más adecuado a la distribución chi-cuadrado que empieza en 0
x = np.linspace(0, chi2.ppf(0.999, df=grados_libertad), 1000)
y = chi2.pdf(x, df=grados_libertad) # PDF de la distribución Chi-cuadrado

plt.figure(figsize=(10, 6))
plt.plot(x, y, label=f'Distribución Chi-cuadrado (df={grados_libertad})')

# Rellenar la región de rechazo (cola izquierda)
x_rechazo = np.linspace(0, valor_critico_chi2, 100)
y_rechazo = chi2.pdf(x_rechazo, df=grados_libertad)
plt.fill_between(x_rechazo, y_rechazo, color='red', alpha=0.5, label=f'Región de Rechazo (α = {alfa})')


# Marcar el valor crítico de Chi^2
plt.axvline(valor_critico_chi2, color='red', linestyle='--', label=f'Valor Crítico Chi^2 = {valor_critico_chi2:.4f}')

# Marcar el estadístico de prueba Chi^2 calculado
plt.axvline(estadistico_chi2, color='blue', linestyle='-', label=f'Estadístico Chi^2 = {estadistico_chi2:.4f}')

# Configuración del gráfico
plt.title('Distribución Chi-cuadrado con Región de Rechazo y Estadístico de Prueba (Prueba de Cola Izquierda)')
plt.xlabel('Valor Chi^2')
plt.ylabel('Densidad de Probabilidad')
plt.legend()
plt.grid(True)
plt.show()

# 7. Tome una decisión estadística y concluya en términos prácticos, explicando el significado del valor p.

# Decisión estadística
print("\nDecisión estadística:")
if estadistico_chi2 < valor_critico_chi2:
    print(f"El estadístico de prueba Chi^2 ({estadistico_chi2:.4f}) es menor que el valor crítico Chi^2 ({valor_critico_chi2:.4f}).")
    print("Por lo tanto, rechazamos la hipótesis nula (H0).")
elif p_valor < alfa:
     print(f"El p-valor ({p_valor:.4f}) es menor que el nivel de significancia ({alfa}).")
     print("Por lo tanto, rechazamos la hipótesis nula (H0).")
else:
    print(f"El estadístico de prueba Chi^2 ({estadistico_chi2:.4f}) no es menor que el valor crítico Chi^2 ({valor_critico_chi2:.4f}).")
    print(f"El p-valor ({p_valor:.4f}) no es menor que el nivel de significancia ({alfa}).")
    print("Por lo tanto, no rechazamos la hipótesis nula (H0).")

# Conclusión en términos prácticos y significado del valor p
print("\nConclusión en términos prácticos:")
if estadistico_chi2 < valor_critico_chi2:
    print(f"Con un nivel de significancia del {alfa*100}%, hay suficiente evidencia para concluir que la varianza poblacional en el grosor de las piezas metálicas es significativamente menor a 0.5 mm^2.")
    print(f"\nSignificado del valor p ({p_valor:.4f}):")
    print(f"El p-valor es la probabilidad de observar una varianza muestral de 0.3 mm^2 o menor en una muestra de 12 piezas, si la varianza poblacional real fuera 0.5 mm^2. Un p-valor pequeño ({p_valor:.4f}) indica que es muy poco probable observar un resultado como el de nuestra muestra si la hipótesis nula fuera cierta. Dado que el p-valor ({p_valor:.4f}) es menor que nuestro nivel de significancia ({alfa}), rechazamos H0.")
else:
    print(f"Con un nivel de significancia del {alfa*100}%, no hay suficiente evidencia para concluir que la varianza poblacional en el grosor de las piezas metálicas es significativamente menor a 0.5 mm^2.")
    print("No podemos rechazar la afirmación de que la varianza poblacional es de 0.5 mm^2.")
    print(f"\nSignificado del valor p ({p_valor:.4f}):")
    print(f"El p-valor es la probabilidad de observar una varianza muestral de 0.3 mm^2 o menor en una muestra de 12 piezas, si la varianza poblacional real fuera 0.5 mm^2. Un p-valor grande ({p_valor:.4f}) indica que es razonablemente probable observar un resultado como el de nuestra muestra si la hipótesis nula fuera cierta. Dado que el p-valor ({p_valor:.4f}) no es menor que nuestro nivel de significancia ({alfa}), no rechazamos H0.")
```
