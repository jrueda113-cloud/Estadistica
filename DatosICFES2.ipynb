```python
import pandas as pd
import numpy as np

# Define the possible cities and socioeconomic strata
cities = ['Bogota', 'Medellin', 'Cali', 'Barranquilla', 'Cartagena']
strata = [1, 2, 3, 4, 5, 6]

# Generate 1000 random scores for Mathematics and English
# Assuming scores are between 0 and 500 for each subject
math_scores = np.random.randint(0, 501, 1000)
english_scores = np.random.randint(0, 501, 1000)

# Generate 1000 random city and strata assignments
assigned_cities = np.random.choice(cities, 1000)
assigned_strata = np.random.choice(strata, 1000)

# Create the DataFrame
icfes_df = pd.DataFrame({
    'Ciudad': assigned_cities,
    'Estrato': assigned_strata,
    'Matematicas': math_scores,
    'Ingles': english_scores
})

# Display the first few rows of the DataFrame
display(icfes_df.head())

import matplotlib.pyplot as plt
import seaborn as sns

# Histogram for Mathematics scores
plt.figure(figsize=(10, 6))
sns.histplot(icfes_df['Matematicas'], kde=True, bins=30)
plt.title('Distribución de Puntajes de Matemáticas')
plt.xlabel('Puntaje de Matemáticas')
plt.ylabel('Frecuencia')
plt.show()

# Histogram for English scores
plt.figure(figsize=(10, 6))
sns.histplot(icfes_df['Ingles'], kde=True, bins=30, color='salmon')
plt.title('Distribución de Puntajes de Inglés')
plt.xlabel('Puntaje de Inglés')
plt.ylabel('Frecuencia')
plt.show()

# Bar plot for Estrato
plt.figure(figsize=(10, 6))
sns.countplot(x='Estrato', data=icfes_df, palette='viridis')
plt.title('Distribución por Estrato Socioeconómico')
plt.xlabel('Estrato')
plt.ylabel('Cantidad')
plt.show()

# Bar plot for Ciudad
plt.figure(figsize=(12, 6))
sns.countplot(x='Ciudad', data=icfes_df, palette='viridis')
plt.title('Distribución por Ciudad')
plt.xlabel('Ciudad')
plt.ylabel('Cantidad')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Muestra aleatoria simple
muestra_aleatoria_simple = icfes_df.sample(n=100, random_state=42)
display(muestra_aleatoria_simple.head())

# Histograms and bar plots for simple random sample
plt.figure(figsize=(10, 6))
sns.histplot(muestra_aleatoria_simple['Matematicas'], kde=True, bins=30)
plt.title('Distribución de Puntajes de Matemáticas (Muestra Simple)')
plt.xlabel('Puntaje de Matemáticas')
plt.ylabel('Frecuencia')
plt.show()

plt.figure(figsize=(10, 6))
sns.histplot(muestra_aleatoria_simple['Ingles'], kde=True, bins=30, color='salmon')
plt.title('Distribución de Puntajes de Inglés (Muestra Simple)')
plt.xlabel('Puntaje de Inglés')
plt.ylabel('Frecuencia')
plt.show()

plt.figure(figsize=(10, 6))
sns.countplot(x='Estrato', data=muestra_aleatoria_simple, palette='viridis')
plt.title('Distribución por Estrato Socioeconómico (Muestra Simple)')
plt.xlabel('Estrato')
plt.ylabel('Cantidad')
plt.show()

plt.figure(figsize=(12, 6))
sns.countplot(x='Ciudad', data=muestra_aleatoria_simple, palette='viridis')
plt.title('Distribución por Ciudad (Muestra Simple)')
plt.xlabel('Ciudad')
plt.ylabel('Cantidad')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()


# Muestra aleatoria estratificada
estrato_proportions = icfes_df['Estrato'].value_counts(normalize=True)
total_sample_size = 100
samples_per_stratum = (estrato_proportions * total_sample_size).round().astype(int)

def stratified_sample(group):
    estrato = group.name
    n_samples = samples_per_stratum.get(estrato, 0)
    if n_samples > 0:
        return group.sample(n=n_samples, random_state=42)
    else:
        return group.sample(n=0, random_state=42) # Handle cases where n_samples is 0

muestra_aleatoria_estratificada = icfes_df.groupby('Estrato').apply(stratified_sample).reset_index(drop=True)

display(muestra_aleatoria_estratificada.head())
display(muestra_aleatoria_estratificada['Estrato'].value_counts())

# Histograms and bar plots for stratified random sample
plt.figure(figsize=(10, 6))
sns.histplot(muestra_aleatoria_estratificada['Matematicas'], kde=True, bins=30)
plt.title('Distribución de Puntajes de Matemáticas (Muestra Estratificada)')
plt.xlabel('Puntaje de Matemáticas')
plt.ylabel('Frecuencia')
plt.show()

plt.figure(figsize=(10, 6))
sns.histplot(muestra_aleatoria_estratificada['Ingles'], kde=True, bins=30, color='salmon')
plt.title('Distribución de Puntajes de Inglés (Muestra Estratificada)')
plt.xlabel('Puntaje de Inglés')
plt.ylabel('Frecuencia')
plt.show()

plt.figure(figsize=(10, 6))
sns.countplot(x='Estrato', data=muestra_aleatoria_estratificada, palette='viridis')
plt.title('Distribución por Estrato Socioeconómico (Muestra Estratificada)')
plt.xlabel('Estrato')
plt.ylabel('Cantidad')
plt.show()

plt.figure(figsize=(12, 6))
sns.countplot(x='Ciudad', data=muestra_aleatoria_estratificada, palette='viridis')
plt.title('Distribución por Ciudad (Muestra Estratificada)')
plt.xlabel('Ciudad')
plt.ylabel('Cantidad')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()


# Muestra sistemática
total_sample_size = 100
n_rows = len(icfes_df)
sampling_interval = n_rows // total_sample_size

import random
start_index = random.randint(0, sampling_interval - 1)

systematic_indices = list(range(start_index, n_rows, sampling_interval))[:total_sample_size]
muestra_sistematica = icfes_df.iloc[systematic_indices]

display(muestra_sistematica.head())

# Histograms and bar plots for systematic sample
plt.figure(figsize=(10, 6))
sns.histplot(muestra_sistematica['Matematicas'], kde=True, bins=30)
plt.title('Distribución de Puntajes de Matemáticas (Muestra Sistemática)')
plt.xlabel('Puntaje de Matemáticas')
plt.ylabel('Frecuencia')
plt.show()

plt.figure(figsize=(10, 6))
sns.histplot(muestra_sistematica['Ingles'], kde=True, bins=30, color='salmon')
plt.title('Distribución de Puntajes de Inglés (Muestra Sistemática)')
plt.xlabel('Puntaje de Inglés')
plt.ylabel('Frecuencia')
plt.show()

plt.figure(figsize=(10, 6))
sns.countplot(x='Estrato', data=muestra_sistematica, palette='viridis')
plt.title('Distribución por Estrato Socioeconómico (Muestra Sistemática)')
plt.xlabel('Estrato')
plt.ylabel('Cantidad')
plt.show()

plt.figure(figsize=(12, 6))
sns.countplot(x='Ciudad', data=muestra_sistematica, palette='viridis')
plt.title('Distribución por Ciudad (Muestra Sistemática)')
plt.xlabel('Ciudad')
plt.ylabel('Cantidad')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()


# Muestra por conglomerados
unique_cities = icfes_df['Ciudad'].unique()
n_cities_to_sample = 3 # You can adjust this number
selected_cities = np.random.choice(unique_cities, size=n_cities_to_sample, replace=False)

muestra_conglomerados = icfes_df[icfes_df['Ciudad'].isin(selected_cities)]

display(muestra_conglomerados.head())

# Histograms and bar plots for cluster sample
plt.figure(figsize=(10, 6))
sns.histplot(muestra_conglomerados['Matematicas'], kde=True, bins=30)
plt.title('Distribución de Puntajes de Matemáticas (Muestra por Conglomerados)')
plt.xlabel('Puntaje de Matemáticas')
plt.ylabel('Frecuencia')
plt.show()

plt.figure(figsize=(10, 6))
sns.histplot(muestra_conglomerados['Ingles'], kde=True, bins=30, color='salmon')
plt.title('Distribución de Puntajes de Inglés (Muestra por Conglomerados)')
plt.xlabel('Puntaje de Inglés')
plt.ylabel('Frecuencia')
plt.show()

plt.figure(figsize=(10, 6))
sns.countplot(x='Estrato', data=muestra_conglomerados, palette='viridis')
plt.title('Distribución por Estrato Socioeconómico (Muestra por Conglomerados)')
plt.xlabel('Estrato')
plt.ylabel('Cantidad')
plt.show()

plt.figure(figsize=(12, 6))
sns.countplot(x='Ciudad', data=muestra_conglomerados, palette='viridis')
plt.title('Distribución por Ciudad (Muestra por Conglomerados)')
plt.xlabel('Ciudad')
plt.ylabel('Cantidad')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
